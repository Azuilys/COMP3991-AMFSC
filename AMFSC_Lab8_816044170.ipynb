## Task 1 — Dimensions and simple bases

For $A\in\mathbb{R}^{M\times N}$ with rank $r$, Cohen (8.25)–(8.27) gives

```python
# Task 1 — complete the TODOs.

def space_dimensions(A, tol=TOL):
    """Return M, N, rank and dimensions of C(A), N(A.T), R(A), N(A)."""
    A = np.asarray(A, dtype=float)
    # TODO: validate A.ndim == 2, then use matrix_rank(A, tol=tol)
    #raise NotImplementedError("Task 1: space_dimensions")
    if A.ndim != 2:
        return DNE
    M, N = A.shape
    r = int(np.linalg.matrix_rank(A, tol = tol))
    return {"M": M, "N": N, "rank": r, "dim C(A)": r, "dim N(A.T)": M-r, "dim R(A)": r, "dim N(A)": N-r}


def independent_column_indices(A, tol=TOL):
    """Greedily keep original columns that increase rank."""
    A = np.asarray(A, dtype=float)
    # TODO: begin with an Mx0 matrix; append each candidate column
    # TODO: keep index j exactly when rank increases
    #raise NotImplementedError("Task 1: independent_column_indices")
    if A.ndim != 2:
        return DNE
    selected = []
    current = np.empty((A.shape[0], 0))
    current_rank = 0
    for j in range(A.shape[1]):
      candidate = np.column_stack((current, A[:, j]))
      candidate_rank = int(np.linalg.matrix_rank(candidate, tol = tol))
      if candidate_rank > current_rank:
        selected.append(j)
        current = candidate
        current_rank = candidate_rank
    return selected


E = np.array([[3., 2., 0.], [1., 1., 4.]])
print("1a E dimensions:", space_dimensions(E))
print("1a E column basis indices:", independent_column_indices(E))
print("1a E row basis indices:", independent_column_indices(E.T))

A = np.array([[1., 2.], [4., 8.]])
col_idx = independent_column_indices(A)
row_idx = independent_column_indices(A.T)
print("1b column basis:\n", A[:, col_idx])
print("1b row basis:\n", A[row_idx, :])
```

```python
# Task 2 — complete the TODOs.

def in_column_space(A, v, tol=TOL):
    A = np.asarray(A, dtype=float)
    v = np.asarray(v, dtype=float).reshape(-1)
    # TODO: validate v.size == A.shape[0], append as column, compare ranks
    #raise NotImplementedError("Task 2: in_column_space")
    if A.ndim != 2 or v.size != A.shape[0]:
        return (DNE, DNE, DNE)
    r1 = int(np.linalg.matrix_rank(A, tol = tol))
    augmented = np.column_stack((A, v))
    r2 = int(np.linalg.matrix_rank(augmented, tol = tol))
    return (r2 == r1, r1, r2)



def in_row_space(A, u, tol=TOL):
    A = np.asarray(A, dtype=float)
    u = np.asarray(u, dtype=float).reshape(-1)
    # TODO: validate u.size == A.shape[1], append as row, compare ranks
    #raise NotImplementedError("Task 2: in_row_space")
    if A.ndim != 2 or u.size != A.shape[1]:
        return (DNE, DNE, DNE)
    r1 = int(np.linalg.matrix_rank(A, tol = tol))
    augmented = np.row_stack((A, u))
    r2 = int(np.linalg.matrix_rank(augmented, tol = tol))
    return (r2 == r1, r1, r2)


A = np.array([[2., 1.], [4., 4.], [0., 0.]])
v = np.array([4., 12., 0.])
w = np.array([4., 1., 1.])
q = np.array([1., 2.])
print("2a v:", in_column_space(A, v))
print("2a w:", in_column_space(A, w))
print("2a q:", in_column_space(A, q))

B = np.array([[0., 1., 1.], [2., 1., 1.], [0., 0., 0.]])
u = np.array([1., 1., 1.])
print("2b u:", in_row_space(B, u))
```

```python
# Task 3 — complete the TODOs.

def is_in_null(A, y, tol=TOL):
    A = np.asarray(A, dtype=float)
    y = np.asarray(y, dtype=float).reshape(-1)
    # TODO: validate y.size == A.shape[1]; compute A@y; compare with zero
    #raise NotImplementedError("Task 3: is_in_null")
    if y.size != A.shape[1]:
        return DNE
    return bool(np.max(np.abs(A@y), initial = 0.0) <= tol)


def is_in_left_null(A, z, tol=TOL):
    A = np.asarray(A, dtype=float)
    z = np.asarray(z, dtype=float).reshape(-1)
    # TODO: validate z.size == A.shape[0]; compute A.T@z; compare with zero
    #raise NotImplementedError("Task 3: is_in_left_null")
    if z.size != A.shape[0]:
        return DNE
    return bool(np.max(np.abs(A.T@z), initial = 0.0) <= tol)


A = np.array([[1., 2.], [4., 8.]])
y = np.array([-2., 1.])
u = np.array([-1., 1.])
print("3a y:", is_in_null(A, y), "A@y =", A @ y)
print("3a u:", is_in_null(A, u), "A@u =", A @ u)

z = np.array([-4., 1.])  # TODO: choose a nonzero vector with A.T @ z = 0
if z is None:
    raise NotImplementedError("Task 3b: left-null vector")
print("3b z:", is_in_left_null(A, z), "A.T@z =", A.T @ z)

E = np.array([[3., 2., 0.], [1., 1., 4.]])
n = np.array([-8., 12., -1.])
print("3c n:", is_in_null(E, n), "E@n =", E @ n)

B = np.array([[1., 1.], [0., 1.]])
# TODO 3d: print B @ A @ y and A @ B @ y
print("3d B @ A @ y =", B @ A @ y)
print("3d A @ B @ y =", A @ B @ y)
```
