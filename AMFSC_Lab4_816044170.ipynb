**1c.** Implement (12.2): For $f(x)=x^2$ on a grid from $0$ to $1$ with $\Delta x=0.25$, build a cumulative approximation to

$$ F(x)=\int_0^x t^2\,dt. $$

Print the array $F$ and its last entry. What does each $F[k]$ approximate? What does the final value $F[-1]$ approximate?


```python
# Task 1 — complete the TODOs.

def cumulative_sum(v):
    # TODO: return a list/array c with c[k] = v[0] + ... + v[k]
    c = []
    total = 0
    for val in v:
        total += val
        c.append(total)
    return np.array(c)


v = np.array([1.0, 3.0, 0.0, -2.0, 4.0])
print("1b  c =", cumulative_sum(v), "  np.cumsum =", np.cumsum(v))

# 1c — (12.2) on f(x)=x**2
dx = 0.25
xs = np.arange(0.0, 1.0 + 1e-12, dx)
f_vals = xs**2  # TODO: xs**2
F = np.cumsum(f_vals) * dx       # TODO: np.cumsum(f_vals) * dx   (or a loop)
if f_vals is None or F is None:
    raise NotImplementedError("Task 1c: set f_vals and F")
print("xs =", xs)
print("F  =", F)
print("last F =", F[-1])
```

**Your answers (Task 1)**

- 1a: c = [1, 4, 4, 2, 6]. Each entry in c is the cumulative sum of the values in v up to that position.
- 1b: The function goes through each value in v, adds it to a running total, and stores the total in the result. It produces the same output as `np.cumsum(v)`.
- 1c: F = [0, 0.015625, 0.078125, 0.21875, 0.46875]. Each value of F approximates the integral of t² from 0 to the corresponding x value. The final value approximates the integral from 0 to 1, which is 1/3. The approximation is too high because it uses right endpoints and x² is an increasing function.


## Task 2 — Riemann left / right / midpoint (questions)

Cohen (13.1)–(13.2) and §13.1.2. $f(x)=x^{2}$ on $[a,b]=[0,1]$, $n=4$.

$$\Delta x=\frac{b-a}{n},\qquad S=\sum_{i=1}^{n} f(x_i)\,\Delta x.$$

**2a.** What is $\Delta x$? List the $n+1$ partition boundaries $x_p$ (13.3).

**2b.** Complete `riemann(f, a, b, n, rule)` for `"left"`, `"right"`, and `"mid"`. Print the three sums for $n=4$.

**2c.** Which of the three is closest to $1/3$ for this increasing $f$?


```python
# Task 2 — complete the TODOs.

def riemann(f, a, b, n, rule="left"):
    dx = (b - a) / n  # TODO: (b - a) / n
    if dx is None:
        raise NotImplementedError("Task 2: set dx")
    if rule == "left":
        xs = a + dx * np.arange(n)  # TODO: a, a+dx, ..., a+(n-1)dx
    elif rule == "right":
        xs =  a + dx * (np.arange(n) + 1) # TODO: a+dx, ..., b
    elif rule == "mid":
        xs =  a + dx * (np.arange(n) + 0.5)  # TODO: midpoints
    else:
        raise ValueError("rule must be left, right, or mid")
    if xs is None:
        raise NotImplementedError("Task 2: set xs for this rule")
    return float(np.sum(f(xs) * dx))


f = lambda t: t**2
a, b, n = 0.0, 1.0, 4
print("dx =", (b - a) / n)
print("left  =", riemann(f, a, b, n, "left"))
print("right =", riemann(f, a, b, n, "right"))
print("mid   =", riemann(f, a, b, n, "mid"))
print("true  =", 1 / 3)
```

**Your answers (Task 2)**

- 2a: The step size is Δx = 0.25. The grid points are xₚ = [0, 0.25, 0.5, 0.75, 1.0].
- 2b:

Using a step size of Δx = 0.25, the estimates are:

Left sum = 0.21875

Right sum = 0.46875

Midpoint sum = 0.328125

The exact value is 1/3 ≈ 0.3333.

- 2c: The midpoint rule gives the closest estimate. Since the function is increasing, the left-endpoint rule underestimates the integral, while the right-endpoint rule overestimates it. The midpoint rule tends to give a more balanced and accurate estimate.

## Task 4 — Net vs total area (questions)

Cohen §13.2. $g(x)=x^{2}-0.5$ on $[-0.5,1.1]$ (the sign-changing example from Ch. 12).

**4a.** Using the midpoint rule and $n=20$, approximate $A_{\mathrm{net}}$ (13.1) and $A_{\mathrm{total}}$ (13.8).

**4c.** Check net area with `sp.integrate` (FTC-2). For total area Cohen uses `sp.integrate(sp.Abs(...), ...)`. Compare.


```python
# Task 4 — complete the TODOs.

g = lambda t: t**2 - 0.5
a, b, n = -0.5, 1.1, 20
dx = (b - a) / n
mids = a + dx * (np.arange(n) + 0.5)

A_net = np.sum(g(mids)) * dx    # TODO: sum g(mids)*dx
A_total = np.sum(np.abs(g(mids))) * dx # TODO: sum abs(g(mids))*dx
if A_net is None or A_total is None:
    raise NotImplementedError("Task 4a: set A_net and A_total")
print("A_net   ~", A_net)
print("A_total ~", A_total)

sx = sp.symbols("sx")
print("sympy net   =", sp.integrate(sx**2 - 0.5, (sx, a, b)))
print("sympy total =", sp.integrate(sp.Abs(sx**2 - 0.5), (sx, a, b)))
```

## Task 7 — Model: recover $C(w)$ from $C'(w)$ (questions)

Same Lab 1 story. Prediction $\hat{y}=wx$ (**weight** $w$). Cost $C(w)=(\hat{y}-y)^{2}$. Data $x=3$, $y=2$.

Last weeks you computed $C'(w)=2x(wx-y)$ and read its sign. FTC-1 run backwards:

$$\int C'(w)\,\mathrm{d}w = C(w)+K.$$

$K$ is the **constant of integration**. Differentiation cannot see it.

**7a.** Write $C'(w)$ (chain rule, Lab 2). Integrate with respect to $w$ (treat $x,y$ as constants). Include $K$.

**7b.** Expand $(wx-y)^{2}$. What must $K$ be?

**7c.** In the cell, use `sp.integrate` on $C'(w)$. Then recover $K$ by matching a known value: $C(1)=1$. Print the recovered $C$ and check $C(2/3)$.

**7d.** One sentence: why did $+K$ cancel in Task 5b but *not* here?


```python
# Task 7 — complete the TODOs.

w = sp.symbols("w", real=True)
x_data, y_actual = 3, 2

Cp = 2 * x_data * (w * x_data - y_actual)  # TODO: 2 * x_data * (w * x_data - y_actual)
if Cp is None:
    raise NotImplementedError("Task 7: set Cp")

F = sp.integrate(Cp, w)
print("int C'(w) dw =", F, "+ K")

# recover K from C(1) = 1
K = 1 - F.subs(w, 1)  # TODO: 1 - F.subs(w, 1)
if K is None:
    raise NotImplementedError("Task 7c: set K so F+K equals C at w=1")
C_rec = sp.simplify(F + K)
print("recovered C(w) =", C_rec)
print("C(1)   =", C_rec.subs(w, 1))
print("C(2/3) =", C_rec.subs(w, sp.Rational(2, 3)))
print("true C =", (w * x_data - y_actual) ** 2)
```

**Your answers (Task 7)**

- 7a: ∫ C′(w) dw = 9w² − 12w + K. The constant K is unknown because differentiation removes constants.
- 7b: (3w − 2)² = 9w² − 12w + 4, so K = 4.
- 7c: Since C(1) = 1, K = 4. Therefore, C(w) = 9w² − 12w + 4. The minimum is C(2/3) = 0.
- 7d: In 5b, K cancels when calculating F(3) − F(0). In Task 7, we’re finding an indefinite integral, so we use C(1) = 1 to find K.
