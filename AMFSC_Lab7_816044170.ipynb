```python
# Task 1 — complete the TODOs.

def product_shape(A, B):
    """Return (M, K) for a valid AB, otherwise DNE."""
    A = np.asarray(A, dtype=float)
    B = np.asarray(B, dtype=float)
    # TODO: require two-dimensional inputs and matching inner dimensions
    #raise NotImplementedError("Task 1: product_shape")
    if A.ndim != 2 or B.ndim != 2 or A.shape[1] != B.shape[0]:
        return DNE
    return (A.shape[0], B.shape[1])


def matmul_element(A, B):
    """Standard AB using the element perspective and three loops."""
    A = np.asarray(A, dtype=float)
    B = np.asarray(B, dtype=float)
    # TODO: call product_shape; return DNE if invalid
    # TODO: allocate C and compute each row-column dot product
    #raise NotImplementedError("Task 1: matmul_element")
    shape = product_shape(A, B)
    if shape == DNE:
        return DNE
    M, K = shape
    N = A.shape[1]
    C = np.zeros((M, K))
    for i in range(M):
        for j in range(K):
            for k in range(N):
                C[i, j] += A[i, k] * B[k, j]
    return C

A_size = np.zeros((2, 4))
B_size = np.zeros((3, 4))
C_size = np.zeros((4, 4))
print("1a AB:", product_shape(A_size, B_size))
print("1a AC:", product_shape(A_size, C_size))
print("1a AB.T:", product_shape(A_size, B_size.T))

A = np.array([[3., 4.], [-1., 2.], [0., 4.]])
B = np.array([[5., 1.], [3., 1.]])
loop_result = matmul_element(A, B)
print("1b loop result:\n", loop_result)
print("1c matches @?", np.allclose(loop_result, A @ B) if not isinstance(loop_result, str) else False)
```

```python
# Task 2 — complete the TODOs.

def outer_loop(a, b):
    a = np.asarray(a, dtype=float).reshape(-1)
    b = np.asarray(b, dtype=float).reshape(-1)
    # TODO: nested loops; output shape (a.size, b.size)
    #raise NotImplementedError("Task 2: outer_loop")
    out = np.zeros((a.size, b.size))
    for i in range(a.size):
        for j in range(b.size):
            out[i, j] = a[i] * b[j]
    return out


def matmul_layer(A, B):
    A = np.asarray(A, dtype=float)
    B = np.asarray(B, dtype=float)
    # TODO: validate; sum outer_loop(A[:, k], B[k, :]) over k
    #raise NotImplementedError("Task 2: matmul_layer")
    shape = product_shape(A, B)
    if shape == DNE:
        return DNE
    C = np.zeros(shape)
    for k in range(A.shape[1]):
        C += outer_loop(A[:, k], B[k, :])
    return C


def matvec(A, v):
    A = np.asarray(A, dtype=float)
    v = np.asarray(v, dtype=float).reshape(-1)
    # TODO: form a weighted sum of the columns of A
    #raise NotImplementedError("Task 2: matvec")
    if A.ndim != 2 or A.shape[1] != v.size:
        return DNE
    out = np.zeros(A.shape[0])
    for k in range(A.shape[1]):
        out += A[:, k] * v[k]
    return out

A = np.array([[3., 4.], [-1., 2.], [0., 4.]])
B = np.array([[5., 1.], [3., 1.]])
print("2a layer result:\n", matmul_layer(A, B))

E = np.array([[3., 2., 0.], [1., 1., 4.]])
lam = np.array([0.5, 0.5, 0.0])
print("2b document d =", matvec(E, lam))

X = np.array([[1., 1.], [2., 1.], [3., 1.]])
w = np.array([0.5, 1.0])
print("2c predictions =", matvec(X, w))
```

### Solutions (tutor) — Task 5

**5a.** All four ranks are $2$. The first two embedding columns are independent, while $e_{\texttt{bus}}=-8e_{\texttt{cat}}+12e_{\texttt{dog}}$, so the third column adds no direction.

**5b.** For $v$, augmentation leaves rank $2$, so $v\in\mathrm{span}(S)$. For $w$, rank rises from $2$ to $3$, so $w\notin\mathrm{span}(S)$; its third component cannot be produced from zeros.

**5c.** For the seeded matrices, $\mathrm{rank}(P)=3$, $\mathrm{rank}(Q)=2$, and $\mathrm{rank}(PQ)=2\le\min\{3,2\}=2$.

**5d.** $R$ has rank $1$ at $\lambda=0$ and rank $2$ for every listed nonzero shift. The diagonal shift removes exact dependence and can improve numerical stability. It is regularisation because it deliberately changes the matrix slightly to obtain better-conditioned computation; a large shift may change the problem too much.

**5e.** With tolerance $10^{-6}$ the tiny direction is treated as zero (rank $1$); with tolerance $10^{-14}$ it is retained (rank $2$). Numerical rank is an estimate based on whether singular values exceed a threshold, so domain knowledge is needed to distinguish signal from noise.

```python
# Solutions (tutor) — Task 5 complete code

def rank_report(A):
    A = np.asarray(A, dtype=float)
    return {
        "rank(A)": int(np.linalg.matrix_rank(A)),
        "rank(A.T)": int(np.linalg.matrix_rank(A.T)),
        "rank(A.T@A)": int(np.linalg.matrix_rank(A.T @ A)),
        "rank(A@A.T)": int(np.linalg.matrix_rank(A @ A.T)),
    }


def in_column_span(S, v, tol=None):
    S = np.asarray(S, dtype=float)
    v = np.asarray(v, dtype=float).reshape(-1, 1)
    if S.ndim != 2 or S.shape[0] != v.shape[0]:
        return (DNE, DNE, DNE)
    augmented = np.column_stack((S, v))
    r1 = int(np.linalg.matrix_rank(S, tol=tol))
    r2 = int(np.linalg.matrix_rank(augmented, tol=tol))
    return (r1 == r2, r1, r2)


def verify_product_rank_law(A, B):
    A = np.asarray(A, dtype=float)
    B = np.asarray(B, dtype=float)
    if A.ndim != 2 or B.ndim != 2 or A.shape[1] != B.shape[0]:
        return DNE
    rA = int(np.linalg.matrix_rank(A))
    rB = int(np.linalg.matrix_rank(B))
    rAB = int(np.linalg.matrix_rank(A @ B))
    bound = min(rA, rB)
    return {"rank(A)": rA, "rank(B)": rB, "rank(AB)": rAB,
            "bound": bound, "law_holds": rAB <= bound}

E = np.array([[3., 2., 0.], [1., 1., 4.]])
print("5a", rank_report(E))
S = np.array([[1., 1.], [1., 7.], [0., 0.]])
v = np.array([1., 2., 0.])
w = np.array([3., 2., 1.])
print("5b v", in_column_span(S, v))
print("5b w", in_column_span(S, w))
rng = np.random.default_rng(17)
P = rng.normal(size=(4, 3))
Q = rng.normal(size=(3, 2))
print("5c", verify_product_rank_law(P, Q))
R = np.array([[1., 2.], [2., 4.]])
lambdas = [0.0, 1e-8, 1e-4, 1e-2, 1.0]
for lam in lambdas:
    shifted = R + lam * np.eye(R.shape[0])
    print(f"5d lambda={lam:g}, rank={np.linalg.matrix_rank(shifted)}")
base = np.array([[1., 3.], [2., 6.], [4., 12.]])
noisy = base.copy()
noisy[-1, -1] += 1e-10
print("5e tol=1e-6:", np.linalg.matrix_rank(noisy, tol=1e-6))
print("5e tol=1e-14:", np.linalg.matrix_rank(noisy, tol=1e-14))
```
