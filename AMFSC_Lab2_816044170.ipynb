## Task 8 — L'Hôpital's rule (questions)

Cohen §7.7. Pronounced ``loh-pee-tahl''. Also spelled L'Hopital / L'Hospital.

If $\lim f/g$ is the **indeterminate** form $0/0$ or $\pm\infty/\infty$, then (when the conditions hold)

$$\lim\frac{f}{g}=\lim\frac{f'}{g'}.$$

This is **not** the quotient rule: $f'/g'\neq (f/g)'$.

Four conditions (§7.7.1): (1) a ratio of functions; (2) the numerator and denominator separately approach values giving a $0/0$ or $\infty/\infty$ indeterminate form; (3) $f'$ and $g'$ exist; (4) $\lim(f'/g')$ exists and $g'\neq 0$.

**8a.** $\lim_{\theta\to 0}\sin(\theta)/\theta$ is $0/0$. Apply L'Hôpital. (Same limit as Lecture 1 with $\varphi$.)

**8b.** $\lim_{x\to\infty}(2x^{2}+4x-7)/(7x^{2}-13x+2)$. Plug-in is $\infty/\infty$. Apply L'Hôpital as many times as needed.
